## 1. Project Description (In Easiest Words)

### ❓ What is the problem?
When people feel sad, stressed, or overwhelmed, they often write posts on Twitter, Reddit, or student forums. 
Traditional software only looks for keywords like `"sad"` or `"depressed"` (which causes lots of mistakes). Humans cannot read millions of posts manually.

### 💡 What is MindPulse AI?
**MindPulse AI** is a smart web app that reads a social media post and answers 3 simple questions:
1. **Is the person in emotional distress?** (Yes / No)
2. **WHY are they distressed?** It puts the reason into one of **6 clear categories**:
   - 🏢 **Jobs/Career**: Burnout, boss issues, exams, unemployment.
   - 💔 **Relationship**: Breakups, fights with family, divorce.
   - 🛑 **Bias/Abuse**: Bullying, online harassment, toxic treatment.
   - 💊 **Medication**: Side effects from psychiatric pills, dosage changes.
   - 🕳️ **Alienation**: Feeling totally alone, invisible, or cut off from the world.
   - ❓ **No Reason**: Feeling low without any clear cause mentioned.
3. **What is the proof?** It copies the **exact words** from the post as evidence (no making things up).

### 🛡️ Ethical Rule (Crucial for Viva/Exams)
> **MindPulse AI NEVER diagnoses people.** It is a research tool, NOT a doctor. If someone is in serious danger, it shows a one-click button to contact 24/7 crisis helplines (like 988).


### 💻 Technologies Used
- **Frontend**: React (for clean user interface) + Vite (super fast server)
- **Backend**: FastAPI (Python web framework to handle requests)
- **AI Brain**: OpenAI (`gpt-4o-mini` for smart reasoning)
- **Knowledge Base (RAG)**: ChromaDB (Vector database storing research guidelines)

A **Level 1 DFD** zooms into the system to show the **main processes**, where data goes, and where it is saved.

### The 4 Basic Symbols:

   - `D1: Research Guidelines Database (ChromaDB)`
   - `D2: Session History Storage (RAM buffer)`
4. ➡️ **Data Flows** (Arrows): Data traveling from one place to another.

### Level 1 DFD Flowchart:
```
           +------------------+
```

## 3. Function Point Counting (FPC / FPA)

### 📐 What is Function Point Counting?
Instead of counting lines of code (which changes if you code in Python vs C++), **Function Points measure what the software does for the user**.
Think of it like **measuring the size of a house by counting bedrooms and bathrooms**, rather than counting individual bricks!

### Step 1: Count the 5 Components (UFP)
1. **External Inputs (EI)**: Ways data enters the app (e.g., submit post, batch submit, clear history).
2. **External Outputs (EO)**: Reports or derived screens created by the app (e.g., result card, batch graphs, JSON export).
3. **External Inquiries (EQ)**: Simple searches / views (e.g., health check, category list, benchmark presets, history view).
4. **Internal Logical Files (ILF)**: Data stored inside the system (e.g., ChromaDB vector store, session history buffer).
5. **External Interface Files (EIF)**: External databases/APIs referenced (e.g., OpenAI embeddings API, OpenAI chat API).


````python
# Function Point Calculation in Python

# 1. Unadjusted Function Points (UFP) Table
# Format: (Component, Count, Weight, Subtotal)
components = [
    ("External Inputs (EI)", 4, 4, 4 * 4),      # 4 inputs with Average weight 4 = 16
    ("External Outputs (EO)", 3, 6, 3 * 6),     # 3 outputs with High/Avg weight = 18
    ("External Inquiries (EQ)", 5, 3, 5 * 3),    # 5 inquiries with Low weight 3 = 15
    ("Internal Logical Files (ILF)", 2, 12, 2 * 12), # 2 stores with High/Avg weight = 24
    ("External Interface Files (EIF)", 2, 9, 2 * 9)   # 2 external APIs with Avg/High = 18
]

UFP = sum(item[3] for item in components)

print("--- Step 1: Unadjusted Function Points (UFP) ---")
for name, count, weight, subtotal in components:
    print(f"{name:30}: {count} items x weight {weight} = {subtotal} FP")
print(f"Total UFP = {UFP}")

# Step 2: Value Adjustment Factor (VAF)
# 14 General System Characteristics (rated 0 to 5)
# Sum of all 14 ratings = 49 (modern web app, security, high usability)
sum_degrees_of_influence = 49
VAF = 0.65 + (0.01 * sum_degrees_of_influence)

# Step 3: Adjusted Function Points (AFP)
AFP = UFP * VAF

print("
--- Step 2 & 3: Adjusted Function Points (AFP) ---")
print(f"Value Adjustment Factor (VAF) = 0.65 + (0.01 * {sum_degrees_of_influence}) = {VAF:.2f}")
print(f"Final Adjusted Function Points (AFP) = {UFP} x {VAF:.2f} = {AFP:.2f} FP")

# Backfiring: Converting FP to Lines of Code (Python ~ 45 LOC/FP)
estimated_loc = AFP * 45
print(f"Estimated Code Size = {AFP:.1f} FP x 45 LOC/FP ≈ {estimated_loc:.0f} Lines of Code")
````

## 4. COCOMO Model (Cost & Effort Estimation)

### ⏱️ What is COCOMO?
**COCOMO** stands for **COnstructive COst MOdel**. It was invented by Barry Boehm to answer three big questions:
1. **How much effort?** (in Person-Months, i.e., 1 person working for 1 month)
2. **How much time?** (in calendar months)
3. **How many people?** (team size)

### Project Type: **Organic Mode**
MindPulse AI is an **Organic** project because:
- Small team ($\le 4$ people)
- Good familiarity with the problem
- Relaxed, academic environment

**Formulas for Organic Mode**:
$$\text{Effort } (E) = 2.4 \times (\text{KLOC})^{1.05} \quad \text{[Person-Months]}$$
$$\text{Development Time } (D) = 2.5 \times (E)^{0.38} \quad \text{[Months]}$$
$$\text{Staff Size } (S) = \frac{E}{D} \quad \text{[Persons]}$$


````python
# COCOMO Calculations in Python

# Size in KLOC (Kilo Lines of Code = Thousands of lines of code)
# Our total project has ~4,500 lines of code = 4.5 KLOC
KLOC = 4.5

# Basic COCOMO Constants for Organic Mode
a = 2.4
b = 1.05
c = 2.5
d = 0.38

# 1. Effort Calculation (Person-Months)
effort = a * (KLOC ** b)

# 2. Time Calculation (Calendar Months)
duration = c * (effort ** d)

# 3. Average Staffing (Persons needed)
team_size = effort / duration

# 4. Total Cost (assuming loaded salary of $4,000 / Rs 1,00,000 per person-month)
cost_usd = effort * 4000

print("--- Basic COCOMO Results ---")
print(f"Code Size (KLOC)   : {KLOC} KLOC (4,500 lines)")
print(f"Effort (E)         : {effort:.2f} Person-Months")
print(f"Development Time (D): {duration:.2f} Months (approx {duration*4:.1f} weeks)")
print(f"Recommended Team   : {team_size:.2f} -> Approx 2 Full-Time Developers")
print(f"Estimated Cost     : ${cost_usd:,.2f} USD")
````

## 5. Critical Path Method (CPM) Project Schedule

### 📅 What is CPM?
When building MindPulse AI, some tasks can be done in parallel, while others have to wait for earlier tasks to finish.
- The **Critical Path** is the **longest continuous chain of dependent tasks**.
- Tasks on the critical path have **ZERO slack (delay tolerance)**. If any critical task is delayed by 1 day, the **entire project launch is delayed by 1 day!**

### Simplified 8-Task Schedule:

| Task ID | Task Name | Needs To Finish First | Days Needed |
| :---: | :--- | :---: | :---: |
| **A** | Requirements & Category Rules | None | 5 days |
| **B** | Database & Vector Guidelines Setup | A | 6 days |
| **C** | Backend API Architecture (FastAPI) | A | 5 days |
| **D** | RAG Search & LLM Inference Logic | B, C | 8 days |
| **E** | Frontend User Interface (React) | A | 6 days |
| **F** | Connect Frontend to Backend (API Client) | D, E | 4 days |
| **G** | End-to-End Testing & Bug Fixing | F | 5 days |
| **H** | Final Documentation & Presentation | G | 3 days |


````python
# Critical Path Method (CPM) Calculation

tasks = {
    'A': {'name': 'Requirements & Rules',      'duration': 5, 'preds': []},
    'B': {'name': 'ChromaDB Guidelines Setup', 'duration': 6, 'preds': ['A']},
    'C': {'name': 'Backend API Architecture',   'duration': 5, 'preds': ['A']},
    'D': {'name': 'RAG & LLM Logic',           'duration': 8, 'preds': ['B', 'C']},
    'E': {'name': 'Frontend UI (React)',        'duration': 6, 'preds': ['A']},
    'F': {'name': 'Connect Frontend + Backend', 'duration': 4, 'preds': ['D', 'E']},
    'G': {'name': 'E2E Testing & Bug Fixing',   'duration': 5, 'preds': ['F']},
    'H': {'name': 'Final Documentation',        'duration': 3, 'preds': ['G']},
}

# Forward Pass (Calculate Early Start and Early Finish)
ES = {}
EF = {}
for tid, task in tasks.items():
    if not task['preds']:
        ES[tid] = 0
    else:
        ES[tid] = max(EF[p] for p in task['preds'])
    EF[tid] = ES[tid] + task['duration']

total_project_days = max(EF.values())

# Backward Pass (Calculate Late Finish and Late Start)
LF = {}
LS = {}
for tid in reversed(list(tasks.keys())):
    successors = [s for s, t in tasks.items() if tid in t['preds']]
    if not successors:
        LF[tid] = total_project_days
    else:
        LF[tid] = min(LS[s] for s in successors)
    LS[tid] = LF[tid] - task['duration']

# Calculate Slack / Float
print(f"{'Task':<6} {'Name':<30} {'Duration':<10} {'ES':<6} {'EF':<6} {'LS':<6} {'LF':<6} {'Slack':<6} {'Critical?'}")
print("-" * 88)
critical_path = []
for tid, task in tasks.items():
    slack = LS[tid] - ES[tid]
    is_crit = "YES (Critical)" if slack == 0 else "No"
    if slack == 0:
        critical_path.append(tid)
    print(f"{tid:<6} {task['name']:<30} {task['duration']:<10} {ES[tid]:<6} {EF[tid]:<6} {LS[tid]:<6} {LF[tid]:<6} {slack:<6} {is_crit}")

print(f"
🎯 Total Project Duration: {total_project_days} Working Days (approx {total_project_days/5:.1f} weeks)")
print(f"🔥 Critical Path: {' -> '.join(critical_path)}")
````

## 6. Agile User Stories (In Simple Words)

In Agile development, requirements are written as **User Stories** from the perspective of the person using the tool:
> **"As a [type of user], I want [a feature], so that [a benefit]."**

---

### 📝 Story 1: Single Post Analysis
- **As a** researcher,
- **I want to** paste a tweet or post and click "Analyze",
- **So that** I can see if there is emotional distress and what category it belongs to.
- **Done When**:
  - The app shows one of the 6 categories (e.g. `Jobs/Career`).
  - The app highlights the exact words from the post as proof.
  - Confidence score is between 0% and 100%.
- **Story Points**: 5

---

### 📝 Story 2: Batch Analysis of Multiple Posts
- **As a** student or researcher with a dataset,
- **I want to** paste up to 25 posts at once,
- **So that** I can analyze them all in one click and see a summary chart.
- **Done When**:
  - All posts are analyzed sequentially or in parallel.
  - A summary progress bar shows percentage distribution (e.g., 40% Jobs, 30% Relationship, 30% None).
- **Story Points**: 8

---

### 📝 Story 3: Emergency Helpline Button
- **As a** user who might be feeling depressed,
- **I want to** see a clear "Crisis Support" button,
- **So that** I can call or text verified 24/7 helplines (like 988) if I need help.
- **Done When**:
  - Clicking the badge opens a popup modal with real telephone numbers (US/Canada, UK, India).
  - Links allow one-click dialing on phone or computer.
- **Story Points**: 2

---

### 📝 Story 4: Clickable Sample Presets
- **As a** tester or professor evaluating the project,
- **I want to** click sample buttons (like "Job Burnout" or "Breakup"),
- **So that** I don't have to type my own test post.
- **Done When**:
  - Clicking any of the 6 preset buttons automatically fills the text box with a curated sample post.
- **Story Points**: 2

---

### 📝 Story 5: Export Analysis History to JSON
- **As a** researcher,
- **I want to** download all analyses done during my session as a JSON file,
- **So that** I can save my experimental results for my research report.
- **Done When**:
  - Clicking "Export JSON" downloads `mindpulse_analysis_history.json` to the user's computer.
- **Story Points**: 3
